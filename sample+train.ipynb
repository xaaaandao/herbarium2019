```python
import os
import pandas as pd
```

```python
local_file = "/mnt/eec07521-c36a-4d2b-9047-0110e7749eae/TAN2019/categories.txt"
```

```python
df = pd.read_csv(local_file)
df.head()
```

```python
df["folder_name"] = df["ScientificName"].str.split(" ").str[0] + "+" + df["ScientificName"].str.split(" ").str[1]
df.head()
```

```python
base_dir = "/mnt/eec07521-c36a-4d2b-9047-0110e7749eae/TAN2019/small-train"
```

```python
import pathlib

for p in pathlib.Path(base_dir).glob("*"):
    src = os.path.join(p)
    mask = df["Category"] == int(p.name)
    label = df.loc[mask, "folder_name"].values[0]
    dst = os.path.join(p.parent, label)
    # os.makedirs(dst, exist_ok=True)
    os.system(f"mv \"{src}\" \"{dst}\"")

    print(f"src {src}")
    print(f"dst {dst}")
```

# Removo espécies que tem mais de uma subespécie

```python
import shutil

for l in os.listdir(base_dir):
    for l2 in os.listdir(os.path.join(base_dir, l)):
        if os.path.isdir(os.path.join(base_dir, l, l2)):
            print(f"removido {os.path.join(base_dir, l)}")
            shutil.rmtree(os.path.join(base_dir, l))
```

```python
for l in os.listdir(base_dir):
    for l2 in os.listdir(os.path.join(base_dir, l)):
        if os.path.isdir(os.path.join(base_dir, l, l2)):
            print(f"diretorio {os.path.join(base_dir, l)} tem pasta")
```
