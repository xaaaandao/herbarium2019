```python
import os
import pathlib
```

```python
t = "train"
min_sample = 50
local_folder = f"/mnt/eec07521-c36a-4d2b-9047-0110e7749eae/TAN2019/small-{t}+labeled"
```

```python
labels = [{"name": p.name, "path": p, "q": len(os.listdir(p))} for p in pathlib.Path(local_folder).glob("*") if p.is_dir() and len(os.listdir(p)) >= min_sample]
labels
```

```python
max_label = max(labels, key=lambda x: x["q"])
min_label = min(labels, key=lambda x: x["q"])
max_label, min_label
```

```python
import shutil
import numpy as np

rng = np.random.default_rng(seed=42) # Usando a seed que vimos anteriormente!
dst = f"/mnt/eec07521-c36a-4d2b-9047-0110e7749eae/TAN2019/small-{t}+labeled-balanced+min={min_sample}"

for l in labels:
    imgs = sorted(os.listdir(l["path"]))
    print(l["path"], len(os.listdir(l["path"])))
    if len(imgs) > min_label["q"]:
        imgs = rng.choice(imgs, size=min_label["q"], replace=False)
    for img in imgs:
        s = os.path.join(l["path"], img)
        d = os.path.join(dst, l["name"])
        os.makedirs(d, exist_ok=True)
        print(f"src: {s}")
        print(f"dst: {d}")
        shutil.copy(s, d)
```
