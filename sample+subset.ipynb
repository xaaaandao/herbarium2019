```python
import pathlib
```

```python
local_folder = "/mnt/eec07521-c36a-4d2b-9047-0110e7749eae/TAN2019/small-train+labeled"
```

```python
genus = [f.name.split("+")[0] for f in pathlib.Path(local_folder).iterdir()]
genus = list(set(genus))
genus
```

```python
count_of_genus = 10
```

```python
import numpy as np

rng = np.random.default_rng(seed=42) # Usando a seed que vimos anteriormente!

choices = rng.choice(genus, size=count_of_genus, replace=False)
print(choices)
```

```python
import os

for p in ["/mnt/eec07521-c36a-4d2b-9047-0110e7749eae/TAN2019/small-train+labeled", "/mnt/eec07521-c36a-4d2b-9047-0110e7749eae/TAN2019/small-validation+labeled"]:
    dst = os.path.join(p+f"+subset+genus={count_of_genus}")
    os.makedirs(dst, exist_ok=True)
    for c in choices:
        cmd = f"cp -r {os.path.join(p, c)}* {dst}"
        os.system(cmd)
```
